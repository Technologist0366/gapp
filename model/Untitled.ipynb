```python
import os
os.environ['DISABLE_MODEL_SOURCE_CHECK'] = 'True'  # keep this if needed

from paddleocr import PaddleOCR
from PIL import Image
import matplotlib.pyplot as plt

# Initialize (downloads models on first run)
ocr = PaddleOCR(use_angle_cls=True, lang='en')   # or 'ch', etc.

# Your image
img_path = 'your_image.jpg'   # replace with real path

# Run OCR
result = ocr.ocr(img_path, cls=True)

# Print detected text (same as before)
for line in result[0]:  # result is usually list of list; [0] for single image
    text, conf = line[1]
    print(f"{text:50} (conf: {conf:.3f})")

# ── Visualization ────────────────────────────────────────────────────────
# result[0] contains the list of detections
image = Image.open(img_path).convert('RGB')

# Extract boxes, texts, scores
boxes   = [line[0] for line in result[0]]
texts   = [line[1][0] for line in result[0]]
scores  = [line[1][1] for line in result[0]]

# Draw using the built-in method (no draw_ocr import needed!)
im_show = ocr.draw_ocr_result(image, boxes, texts, scores)   # ← this is the new way

# Or save it directly:
# ocr.draw_ocr_result(image, boxes, texts, scores, save_path='result.jpg')

# Show in notebook
im_show = Image.fromarray(im_show)   # if needed
plt.figure(figsize=(12, 12))
plt.imshow(im_show)
plt.axis('off')
plt.show()
```
