```python
import os
os.environ['DISABLE_MODEL_SOURCE_CHECK'] = 'True'  # keep if helpful

from paddleocr import PaddleOCR
import matplotlib.pyplot as plt  # already installed
from PIL import Image
import glob  # for listing images if needed

# Initialize with 3.x parameters (use_textline_orientation replaces old cls/use_angle_cls)
ocr = PaddleOCR(
    lang='en',
    use_textline_orientation=True,       # enables text-line angle/orientation classification
    use_doc_orientation_classify=False,  # optional: document-level orientation (set True if needed)
    use_doc_unwarping=False             # optional: deskew/unwarp (set True for curved docs)
)

# Your directory (will process ALL images inside it)
img_dir = '/home/mehit/Documents/complianceassets/model/dataset/kdreceipts/'

# Run prediction on the directory (batch processes all images)
results = ocr.predict(img_dir)

# Output folder for visualized results (creates if not exists)
output_dir = 'ocr_results'
os.makedirs(output_dir, exist_ok=True)

print("Processing images in directory...")

# Iterate over each image's result
for idx, res in enumerate(results):
    # res is a dict-like object for one image
    print(f"\n--- Image {idx+1} ---")
    
    # Extract text + confidence (adjust keys if needed based on your res structure)
    for i, (box, (text, conf)) in enumerate(zip(res['dt_polys'], zip(res['rec_texts'], res['rec_scores']))):
        print(f"  {text:50} (conf: {conf:.3f})")
    
    # Built-in visualization: save annotated image (boxes + text overlaid)
    res.save_to_img(output_dir)  # saves as e.g., ocr_results/image_name_ocr.jpg
    
    # Optional: show first image in notebook (for quick preview)
    if idx == 0:  # just preview the first one
        annotated_path = glob.glob(f"{output_dir}/*.jpg")[0]  # get the latest saved
        img_show = Image.open(annotated_path)
        plt.figure(figsize=(12, 12))
        plt.imshow(img_show)
        plt.title(f"PaddleOCR Result - First Image")
        plt.axis('off')
        plt.show()
```
