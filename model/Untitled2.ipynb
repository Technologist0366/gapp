```python
import os
os.environ['DISABLE_MODEL_SOURCE_CHECK'] = 'True'

from paddleocr import PaddleOCR
import matplotlib.pyplot as plt
from PIL import Image
import glob
import os

# Optional: disable oneDNN if still issues
# os.environ['FLAGS_use_mkldnn'] = '0'

ocr = PaddleOCR(
    lang='en',
    use_textline_orientation=True,       # for angle cls
)

img_dir = '/home/mehit/Documents/complianceassets/model/dataset/kdreceipts/'

results = ocr.predict(img_dir)  # processes all images in dir

output_dir = 'ocr_results'
os.makedirs(output_dir, exist_ok=True)

for idx, res in enumerate(results):
    print(f"\nImage {idx+1}:")
    for text, conf in zip(res['rec_texts'], res['rec_scores']):
        print(f"  {text:50} (conf: {conf:.3f})")
    
    res.save_to_img(output_dir)  # saves annotated image
```
